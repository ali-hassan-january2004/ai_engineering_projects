```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Task 1: Create or load the dataset with Pandas
data = {
    "name": ["Ali", "Sara", "Usman", "Hina", "Zain", "Ayesha", "Bilal"],
    "age": [23, 22, 24, 21, 23, 22, 25],
    "study_hours": [4.5, 6.0, 2.5, 5.0, 3.0, 7.5, 1.5],
    "attendance": [88, 95, 60, 90, 70, 98, 50],
    "assignments": [82, 94, 55, 88, 65, 96, 40],
    "final_score": [85, 92, 48, 88, 58, 96, 35]
}

df = pd.DataFrame(data)

# Task 2: Display the first five rows
print("--- First 5 Rows ---")
print(df.head())

# Task 3: Display shape, data types, and descriptive statistics
print("\n--- Dataset Shape ---")
print(df.shape)

print("\n--- Data Types & Info ---")
print(df.info())

print("\n--- Summary Statistics ---")
print(df.describe())

# Task 4: Check for missing values
print("\n--- Missing Values Count ---")
print(df.isnull().sum())

# Task 5: Calculate the average final score
avg_score = df["final_score"].mean()
print(f"Average Final Score: {avg_score:.2f}")

# Task 6: Find the student with the highest final score
top_student_row = df[df["final_score"] == df["final_score"].max()]
top_student_name = top_student_row["name"].values[0]
top_student_score = top_student_row["final_score"].values[0]
print(f"Top Performing Student: {top_student_name} with a score of {top_student_score}")

# Task 7: Create a 'passed' column where final_score >= 50
df["passed"] = df["final_score"] >= 50
print("\n--- Updated DataFrame with 'passed' Column ---")
print(df[["name", "final_score", "passed"]])

# Task 8: Calculate average final score for students with at least 4 study hours
high_study_df = df[df["study_hours"] >= 4.0]
avg_score_high_study = high_study_df["final_score"].mean()
print(f"\nAverage Final Score for students studying >= 4 hours: {avg_score_high_study:.2f}")
```

```python
# Task 9: Histogram of final scores
plt.figure(figsize=(8, 4))
plt.hist(df["final_score"], bins=5, color='skyblue', edgecolor='black')
plt.title("Distribution of Final Scores")
plt.xlabel("Final Score")
plt.ylabel("Number of Students")
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Task 10: Scatter plot of study_hours versus final_score
plt.figure(figsize=(8, 4))
plt.scatter(df["study_hours"], df["final_score"], color='darkblue', s=100)
plt.title("Study Hours vs. Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.grid(True)
plt.show()
```
