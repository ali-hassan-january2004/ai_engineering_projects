```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Task 4: Create the Student Performance dataset
data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Ali", "Sara", "Usman", "Hina", "Zain", "Ayesha", "Bilal", "Tariq"],
    "age": [23, 22, 24, 21, 23, 22, 25, 23],
    "study_hours": [4.5, 6.0, 2.5, 5.0, 3.0, 7.5, 1.5, 5.5],
    "attendance": [88, 95, 60, 90, 70, 98, 50, 85],
    "assignments": [82, 94, 55, 88, 65, 96, 40, 80],
    "final_score": [85, 92, 48, 88, 58, 96, 35, 82]
}

df = pd.DataFrame(data)
df.head()
```

```python
# Task 5: Required Pandas Analysis
print("--- Dataset Summary Info ---")
print(df.info())

print("\n--- Descriptive Statistics ---")
print(df.describe())

print("\n--- Missing Values Check ---")
print(df.isnull().sum())

# Filter top performers and add derived column
df["passed"] = df["final_score"] >= 50
high_performers = df[df["final_score"] >= 80]

print("\n--- Students Scoring 80+ ---")
print(high_performers[["name", "study_hours", "final_score"]])
```

```python
# Task 6: NumPy Numerical Operations (3 Operations)

# 1. Convert pandas column to numpy array and calculate percentile
scores_array = df["final_score"].to_numpy()
p75 = np.percentile(scores_array, 75)

# 2. Standard deviation of study hours
study_std = np.std(df["study_hours"].to_numpy())

# 3. Conditional extraction / masking via numpy
failing_scores = np.extract(scores_array < 50, scores_array)

print(f"75th Percentile Score: {p75:.2f}")
print(f"Study Hours Std Dev: {study_std:.2f}")
print(f"Failing Scores Array: {failing_scores}")
```

```python
# Task 7: Two Meaningful Visualizations

plt.figure(figsize=(12, 5))

# Plot 1: Histogram of Final Scores
plt.subplot(1, 2, 1)
plt.hist(df["final_score"], bins=5, color='teal', edgecolor='black')
plt.title("Distribution of Final Scores")
plt.xlabel("Score Range")
plt.ylabel("Number of Students")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Plot 2: Scatter Plot (Study Hours vs. Final Score)
plt.subplot(1, 2, 2)
plt.scatter(df["study_hours"], df["final_score"], color='crimson', s=90)
plt.title("Study Hours vs. Final Score")
plt.xlabel("Study Hours per Week")
plt.ylabel("Final Score")
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
## Key Findings & Insights

1. **Study Hours Impact:** There is a strong positive correlation between weekly study hours and final scores. Students studying over 5 hours consistently achieved scores above 80.
2. **Class Pass Rate:** 75% of the students (6 out of 8) successfully passed the assessment with a score \(\ge 50\).
3. **Data Quality:** The dataset contained no missing or null values across all 8 student entries.
4. **Outlier Identification:** Students with attendance below 60% (Usman and Bilal) failed the course, highlighting attendance as a key performance indicator.
```
