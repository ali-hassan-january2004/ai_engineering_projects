```python
import numpy as np  
# 'import' loads the library. 'as np' lets you write 'np' instead of typing 'numpy' every time.

x = np.array([10, 20, 30, 40]) 
# Creates a 1D list of numbers optimized for fast math operations.

print(x)        # Output: [10 20 30 40]
print(x.shape)  # Output: (4,) — Tells you the dimensions (4 elements in 1 dimension).
print(x.dtype)  # Output: int64 or int32 — Shows the data type (integers).

matrix = np.array([[1, 2, 3], [4, 5, 6]]) 
# Creates a 2D grid (2 rows, 3 columns).

print(matrix.shape) # Output: (2, 3) — 2 rows, 3 columns.
```

```python
# Vectorized Operations (Math performed on every single element automatically)
x = np.array([1, 2, 3, 4]) 
print(x * 10)  # Output: [10 20 30 40] — Multiplies every item by 10.
print(x + 5)   # Output: [6 7 8 9]     — Adds 5 to every item.

# Useful Helper Functions
np.zeros((2, 3))    # Creates a 2x3 matrix full of zeros.
np.ones((2, 3))     # Creates a 2x3 matrix full of ones.
np.arange(0, 10, 2) # Generates numbers from 0 up to (not including) 10, stepping by 2: [0, 2, 4, 6, 8].
np.mean(x)          # Calculates the average of array x.
np.std(x)           # Calculates the standard deviation of array x.
np.min(x) / np.max(x) # Returns the lowest/highest value in x.
x.reshape(2, 2)     # Changes a 1D array of 4 items into a 2x2 matrix.
```

```python
import pandas as pd  
# Loads pandas with the standard shorthand 'pd'.

# Create a dictionary of data (column name -> list of values)
data = { 
    "name": ["Ali", "Sara", "Usman", "Hina"], 
    "age": [23, 24, 22, 25], 
    "score": [82, 91, 76, 88] 
} 

df = pd.DataFrame(data) # Converts the dictionary into a formatted table.

print(df.head())    # Displays the first 5 rows of the table.
print(df.shape)     # Displays the dimensions: (4, 3) -> 4 rows, 3 columns.
print(df.info())    # Summary of column names, data types, and non-null values.
print(df.describe())# Statistical summary (mean, min, max, std) for numerical columns.
```

```python
import pandas as pd

data = { 
    "name": ["Ali", "Sara", "Usman", "Hina"], 
    "age": [23, 24, 22, 25], 
    "score": [82, 91, 76, 88] 
} 

df = pd.DataFrame(data)

# Selecting and Filtering
print(df["score"])             # Selects a single column ("score").
print(df[["name", "score"]])   # Selects multiple columns ("name" and "score").

high_scores = df[df["score"] >= 85] 
# Filters rows where 'score' is 85 or higher (Sara and Hina).
print(high_scores)

# Creating a Derived Column
df["passed"] = df["score"] >= 50 
# Adds a new column 'passed' containing True/False based on the score.
print(df)
```
