```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Task 1: Load the Week 1 Student Performance Dataset
data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Ali", "Sara", "Usman", "Hina", "Zain", "Ayesha", "Bilal", "Tariq"],
    "study_hours": [4.5, 6.0, 2.5, 5.0, 3.0, 7.5, 1.5, 5.5],
    "final_score": [85, 92, 48, 88, 58, 96, 35, 82]
}

df = pd.DataFrame(data)
df.head()
```

```python
# Task 2: Statistics for final_score
scores = df["final_score"].to_numpy()
print("--- Final Score Statistics ---")
print(f"Mean:               {np.mean(scores):.2f}")
print(f"Median:             {np.median(scores):.2f}")
print(f"Variance:           {np.var(scores, ddof=1):.2f}")
print(f"Standard Deviation: {np.std(scores, ddof=1):.2f}")

# Task 3: Statistics for study_hours
hours = df["study_hours"].to_numpy()
print("\n--- Study Hours Statistics ---")
print(f"Mean:               {np.mean(hours):.2f}")
print(f"Median:             {np.median(hours):.2f}")
print(f"Variance:           {np.var(hours, ddof=1):.2f}")
print(f"Standard Deviation: {np.std(hours, ddof=1):.2f}")
```

```python
# Task 4: Calculate Pearson correlation coefficient
corr_matrix = np.corrcoef(hours, scores)
correlation = corr_matrix[0, 1]
print(f"Correlation between study_hours and final_score: {correlation:.4f}")

# Task 5: Scatter plot vs. Correlation comparison
plt.figure(figsize=(7, 4))
plt.scatter(hours, scores, color="crimson", s=80, label="Student Data")
plt.title(f"Study Hours vs. Final Score (r = {correlation:.2f})")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```

```python
# Task 6: Implement linear hypothesis function y_hat = w*x + b
def predict(x, w, b):
    return w * x + b

# Task 7: Implement Mean Squared Error (MSE) loss function
def compute_mse(y_true, y_pred):
    n = len(y_true)
    return (1 / n) * np.sum((y_true - y_pred) ** 2)

# Experimenting with arbitrary initial weights
w_initial = 10.0
b_initial = 30.0

y_preds = predict(hours, w_initial, b_initial)
initial_loss = compute_mse(scores, y_preds)

print(f"Initial Predictions: {y_preds}")
print(f"Initial MSE Loss: {initial_loss:.2f}")
```

```python
# Task 8: Gradient Descent optimization on simple f(x) = x^2
x = 10.0
alpha = 0.1
epochs = 20

print("--- Gradient Descent Optimization ---")
for step in range(1, epochs + 1):
    grad = 2 * x  # Derivative df/dx
    x = x - alpha * grad  # Update rule
    print(f"Step {step:2d} | Gradient: {grad:6.2f} | x: {x:6.4f}")
```

## Connecting Mathematics to Machine Learning Training

The mathematical components implemented in this lab form the core pipeline of supervised machine learning:

1. **Hypothesis Function (\(\hat{y} = wx + b\)):** Converts input features (\(x\)) into continuous target predictions (\(\hat{y}\)) controlled by parameters \(w\) (weight) and \(b\) (bias).
2. **Loss Function (\(\text{MSE} = \frac{1}{n} \sum (y_i - \hat{y}_i)^2\)):** Measures total model prediction error by measuring squared distances from actual targets.
3. **Derivatives (\(\nabla L\)):** Compute the slope and direction of error change relative to each parameter.
4. **Gradient Descent (\(\theta \leftarrow \theta - \alpha \nabla L\)):** Iteratively adjusts parameters opposite the gradient direction by step size \(\alpha\), systematically minimizing overall model loss until optimal convergence is achieved.
