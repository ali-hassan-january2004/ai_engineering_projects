```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Dot Product: (1*4) + (2*5) + (3*6) = 4 + 10 + 18 = 32
print("dot:", np.dot(a, b))

# Euclidean Norm (Magnitude): sqrt(1^2 + 2^2 + 3^2) = sqrt(14) ≈ 3.74
print("norm:", np.linalg.norm(a))

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Transpose: Flips rows into columns -> [[1, 3], [2, 4]]
print("transpose:\n", A.T)

# Matrix Multiplication (A @ B): Row-by-column dot products
# [[(1*5 + 2*7), (1*6 + 2*8)], [(3*5 + 4*7), (3*6 + 4*8)]] -> [[19, 22], [43, 50]]
print("product:\n", A @ B)

# Mean of all matrix elements: (1+2+3+4)/4 = 2.5
print("mean:", np.mean(A))

# Standard Deviation of all matrix elements ≈ 1.118
print("std:", np.std(A))
```

```python
x = 10.0
learning_rate = 0.1

# Finds the minimum of f(x) = x^2 (where derivative is 2*x)
for step in range(20):
    gradient = 2 * x  # Derivative f'(x)
    x = x - learning_rate * gradient  # Parameter update step
    print(step, x)  # Prints step number and x moving closer to 0
```

```python
import numpy as np

scores = np.array([55, 60, 68, 72, 80, 85, 90, 95])

print("mean:", np.mean(scores))      # Average: 75.625
print("median:", np.median(scores))  # Middle value: 76.0
print("variance:", np.var(scores))   # Squared spread around mean: 171.234
print("std:", np.std(scores))        # Spread in score units: 13.085
print("min:", np.min(scores))        # Lowest score: 55
print("max:", np.max(scores))        # Highest score: 95
```
